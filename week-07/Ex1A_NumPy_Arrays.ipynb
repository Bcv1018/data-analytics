## Looping Through Arrays

```python
arrA = np.array([[11,12,13],[14,15,16]])
print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y*2)
```

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA,arrA), axis=1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1,joined_array_1))
print(stacked_array_1)
```

```python
stacked_array_2 = np.stack((joined_array_1,joined_array_2))
# The error message I get is a valueError all input arrays must have the same shape
```

```python
another_joined_array = np.concatenate((joined_array_1,arrA))
print(another_joined_array)
```

```python
another_2 = np.concatenate((joined_array_1,arrA),axis=1)
# The arrays are different sizes and need to match in size in order to concatenate 
```

```python
arr_hstack = np.hstack((arrA,arrA)) # stacks each 1D arrays horizontally 
arr_vstack = np.vstack((arrA,arrA)) # stacks each 1D arrays on to top of each vertically
arr_dstack = np.dstack((arrA,arrA)) # stacks each 1D arrays in a third dimension for depth

print(arr_hstack) 
print(f'\n{arr_vstack}') 
print(f'\n{arr_dstack}')
```

## Splitting Arrays

```python
splitsky = np.split(joined_array_1, 2)
print(splitsky) 
# it split it into 2 arrays with 2 dimensions each array
```

```python
spliterino = np.array_split(joined_array_1, 3)
print(spliterino)
# it split it into 3 arrays with 2 dimensions each array
```

## Sorting and Filtering Arrays

```python
arrB = np.array(([[7,6],[13,5],[7,3]]))
arrB_sorted = np.sort(arrB)
print(arrB_sorted)
```

```python
print(arrB)

z = arrB < 7
arrC = arrB[z]
print(f'\n{arrC}')
```

```python
print(arrB)

y = arrB >= 7
arrD = arrB[y]

print(f'\n{arrD}')
```
