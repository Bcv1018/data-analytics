# Exercise 3.A Getting Familiar with Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
# Creating two variables that contain lists with x and y plot points

x = [1,2,3,4,5]
y = [2,3,5,6,11]
```

```python
# Creating a graph from two different lists

plt.plot(x,y)
plt.show()
```

```python
# Adding markers to the graph

plt.plot(x,y, marker='o')
plt.show()
```

```python
# Working with xlabel, ylabel, and title

plt.plot(x,y, marker='o')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Example Graph')
plt.show()
```

## Styling Plots

```python
# Working with changing the line style and changing the color of the line

# Creating and modifying the graph
plt.plot(x,y, marker='o',ls= 'dashed',color= 'r')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Example Graph')


plt.show()
```

```python
# Changing the marker style to a square, changing the marker size , and changing the marker color

# Creating and modifying the graph
plt.plot(x,y, marker='s',ls= 'dashed',color= 'r',markersize=10,mfc='g') # changed the marker style to a square, size to 10, and color to green
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Example Graph')


plt.show()
```

```python
# Experimenting with two different marker styles and color

# Creating and modifying the graph
plt.plot(x,y, marker='*',ls= 'dashed',color= 'r',markersize=30,mfc='k') # changed the marker style to a star, size to 30, and color to black 
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Example Graph')


plt.show()
```

```python
# Experimenting with two different marker styles and color

# Creating and modifying the graph
plt.plot(x,y, marker='d',ls= 'dashed',color= 'r',markersize=5,mfc='m') # changed the marker style to a thin diamond, size to 30, and color to black 
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Example Graph')


plt.show()
```

# Exercise 3.B More plotting practice

## Advanced Plotting Techniques

```python
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]

plt.bar(products,sales)
plt.show()
```

```python
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]

plt.bar(products,sales, color='skyblue',width=.5)
plt.title('Product Sales')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()
```

```python
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x,y)

plt.show()
```

```python
s = np.random.rand(50) * 100
plt.scatter(x,y,s = s,c=y, cmap='plasma')

plt.show()
```

## Subplots

```python
x = np.linspace(0,10,100)
y = np.sin(x)
thou = np.random.randn(1000)

# Create subplot with histogram and line chart
fig,(ax1,ax2) = plt.subplots(1,2)

ax1.plot(x,y, c='red')
ax1.set_title('Sine Wave')
ax1.set_xlabel('X axis')
ax1.set_ylabel('Y axis')

ax2.hist(thou)
ax2.set_title('1000 Ran Numbers')
ax2.set_xlabel('X axis')
ax2.set_ylabel('Y axis')

plt.tight_layout()
plt.show()
```
