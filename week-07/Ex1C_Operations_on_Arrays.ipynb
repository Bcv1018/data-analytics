## Grocery Problem

```python
# Arrays are vegetables,fruits,boxes of snacks, and beverages
bag1 = np.array([2,3,4,0])
bag2 = np.array([4,0,1,2])

print(bag1)
print(bag2)
```

```python
# Write an expression to calculate the total number of items in each bag, and print the result in the form of a sentence

print(f'The first bag contains {np.sum(bag1)} items, and the second bag contains {np.sum(bag2)} items.')
```

## Teamwork

```python
# Arrays for each member will be ordered as Monday-Friday, monday being index 0 etc etc
ali = np.array([5,4,6,4,2])
ben = np.array([0,5,5,3,4])
carla = np.array([3,6,0,3,4])
```

```python
# Compile the values reported by Ali, Ben, and Carla into a single array containing all door-knocks for the week, as all_knocks. Print the resulting array.
all_knocks = np.array([ali,ben,carla])
print(all_knocks)
```

```python
# Use all_knocks to calculate the total number of doors that each person knocked on for the week, and also the total number of doors that the group collectively knocked on. Print the results.

# Total amount of doors knocks as a group
total_doors_knocks = np.sum(all_knocks)
print(f'The total amount of doors knocked for the week is {total_doors_knocks} doors')

# Total amount doors knocked per person
door_per_person = np.sum(all_knocks, axis=1)
print(f'Ali knocked {door_per_person[0]} doors\nBen knocked {door_per_person[1]}\nCarla knocked {door_per_person[2]} doors')

```

```python
# what is the rolling total of estimated votes that your group was able to turn out, day by day for last week?
daily_knocks = np.sum(all_knocks,axis=0)
# print(daily_knocks)

rolling_total = np.cumsum(daily_knocks) * .5
print(f"Estimated votes for each day:\nMonday's votes: {rolling_total[0]:.0f}\nTuesday's votes: {rolling_total[1]:.0f}\nWednesday's votes: {rolling_total[2]:.0f}\nThursday's votes: {rolling_total[3]:.0f}\nFriday's votes:{rolling_total[4]:.0f}")
```
