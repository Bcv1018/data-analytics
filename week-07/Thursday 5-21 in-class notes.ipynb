```python
nw['LineTotal'] = nw['UnitPrice'] * nw['Quantity']
nw
```

```python
prod_by_rev = nw.groupby('ProductName')[['Quantity','LineTotal']].sum()
prod_by_rev.columns = ['Total Qty Sold','Total Product Revenue']

prod_by_rev
```
