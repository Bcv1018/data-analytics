### Working with datetime functions

```python
import datetime
```

## Today

```python
today = datetime.datetime.now() # current date and time
today # displays today varible giving us the current date and time when its ran
```

```python
day_of_week = today.strftime("%A") # gives us the name of the day
month = today.strftime("%m") # gives us the month in MM format
day = today.strftime("%d") # gives us the day in DD format
year = today.strftime("%Y") # gives us the year with century in YYYY format

print(f'Today is {day_of_week}, {month}/{day}/{year}') # shows today's date but formatted nicely in MM/DD/YYYY format and states day of the week
```

## My Birthday

```python
birthday = datetime.date(2003,11,8) # shows my birthday date unformated
birthday = birthday.strftime("%x") # formats my birthday into locale short date mm/dd/yy
print(f'My birthday is {birthday}') # shows my birthday but formatted nicely in MM/DD/

# I had birthday unformatted and formatted in two different cells ran fine once then when I ran it again it gave me this error
# AttributeError                            Traceback (most recent call last)
# Cell In[41], line 1
# ----> 1 birthday = birthday.strftime("%x")
#       2 print(f'My birthday is {birthday}') AttributeError: 'str' object has no attribute 'strftime'
# But when I put them in the same cell it works fine!
```

## Next 90 Days

```python
ninety_d = today + datetime.timedelta(days=90) # adds current date with 90 days
ninety_d # check calculation is working right
```

```python
ninety_day = ninety_d.strftime("%d") # formats 90th day in DD format
ninety_month = ninety_d.strftime("%m") # formats 90th day month in MM format
ninety_year = ninety_d.strftime("%Y") # formats 90th day year in YYYY format

print(f"90 days from today is {ninety_month}/{ninety_day}/{ninety_year}") # shows 90 days from today but formatted nicely in MM/DD/YYYY format
```

## Time for Dinner

```python
dinner_time = datetime.time(5,30) # Shows hour and minute
dinner_time # checks if dinner_time is showing right time
```

```python
hour = dinner_time.strftime("%I") # Formats hour into 12 hour zero padded decimal number
minute = dinner_time.strftime("%M") # Formats minutes into as a zero padded decimal number

print(f"Let's meet for dinner at {hour}:{minute} PM") # shows dinner time in HH:MM PM format
```
