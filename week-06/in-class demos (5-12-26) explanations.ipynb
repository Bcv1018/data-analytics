### Using random library

```python
import random
```

```python
a = random.randint(1,100) # will return a random integer from the range 1-100
a
```

```python
print(a)
```

```python
b = []

for i in range(10): 
    b.append(random.randint(1,100)) # will run random numbers from the range 1-100 10 ten times and put it in the empty list b

print(b)
```

```python
random.choice(b) # only returns one option
```

```python
random.choices(b) # only returns one item in a list
```

```python
# .choices() using optional parameters

random.choices(b, weights=[1,2,1,2,1,2,1,2,6,7,], k=10) # weights gives probablity chances of how much a value in the list will appear 
# k is just how many values it will return
```

```python
random.choices(b, weights=[1,2,1,2,1,2,1,2,6,7,], k=10)
```

```python
print(b)
```

```python
random.shuffle(b) # reorders the items in the list
```

```python
print(b)
```

```python
# no weighting componet to .sample()
# sample could not be larger than the set size example

random.sample(b, 6) # (b, k=6) dont need to write k as it already knows since it is only 2 different values in sample
```

## Custome Function
### working with def keyword

```python
# very simple example of a function:

def pizza():
    return "Yay pizza"
```

```python
pizza()
```

```python
def pizza_v_burritos(food_name):
    if food_name == 'pizza':
        msg = 'Yay pizza!'
    elif food_name == 'burritos':
        msg = 'Yay burritos!'
    else:
        msg = 'boooooooooo'
    return msg
```

```python
pizza_v_burritos('burritos')
```

```python
pizza_v_burritos('pizza')
```

```python
pizza_v_burritos('liver and onions')
```

```python
# pizza_v_burritos('pizza', 'burritos') 
# Will result in error -- can only take 1 argument not 2 or more
```

```python
#Example of a function with 2 parameters

def place_order(item, qty):
    return f'You ordered {qty} of item {item}'
```

```python
place_order('pizza', 2)
```

```python
place_order('hockey puck', 67)
```

```python
place_order(input('What would you like to order (pizza or burrito)? '), input('How many? '))
```

```python
place_order(qty="7", item="burritos")
```

```python
def tax_calc(rate, amount, exempt='N'):
    if exempt != "N":
        tax_amt = 0
    else:
        tax_amt = rate * amount
    return tax_amt
```

```python
tax_calc(.08, 122.45)
```

```python
tax_calc(.075, 58, "exempt")
```

## working with an arbitrary number of arguments

```python
def greeting_kids(*args):
    return args
```

```python
greet = greeting_kids('Henry','Alice')

greet
```

```python
def send_greetings(whomst):
    for name in whomst:
        print(f'Hi, {name}')
```
