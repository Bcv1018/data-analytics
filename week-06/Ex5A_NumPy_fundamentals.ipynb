## Array Basics

```python
fruits = ['apple','banana','cherry','date']
```

```python
arr_1 = np.array(fruits)
print(arr_1)
```

```python
print(arr_1.dtype)
# The data type that is returned is U6 which the U is unicorn string and 6 is the max length of the strings in the array
```

```python
arr_2 = ([10,11,12],[13,14,15])
```

```python
arr_2 = np.array(arr_2)
print(arr_2)
```

```python
print(arr_2.T) # It does not alter the value stored in arr_2
```

```python
print(arr_2)
```

```python
arr_3 = ([1,3,5],[2,4,6])
arr_3 = np.array(arr_3)
print(arr_3)
```

```python
arr_4 = np.array([arr_3[0,2],arr_3[1,1]])
print(arr_4)
```
