## Lab 3. Employee data

### This lab exercise makes updates to information extracted from an employee database.

```python
hr_list = [
    ('0123', 'HR', 'Rebecca Yang', '69000'),
    ('0121', 'IT', 'Mark Blick', '67000'),
    ('0068', 'IT', 'Kahna Larsen', '112000'),
    ('0234', 'OPS', 'Jim Smith', '54000')
]
```

```python
# Updates Mark's last name by replacing the old tuple with a new one with the updated name
hr_list[1] = ('0121', 'IT', 'Mark Blick-Hawley', '67000')
```

```python
# Updated list
hr_list
# Prints updated list
print(hr_list)
```

```python
# Updates Jims possition by changing his dep code and salary
hr_list[3] = ('0234', 'CS', 'Jim Smith', '60000')
```

```python
# updates list
hr_list
# prints updated list
print(hr_list)
```
