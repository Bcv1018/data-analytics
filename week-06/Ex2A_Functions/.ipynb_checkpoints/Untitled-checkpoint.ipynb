# Exercise 2.A Using built-in functions
## Lab 1. Working with names - parsing strings

```python
test_name = 'Jackie Chan'
```

```python
def parse_and_display(name):
    for i in range(len(name)):
        if name [i] == ' ':
            first = name [:i]
            last = name[i+1:]
            print(f"Full name : {name}")
            print(f"First name : {first}")
            print(f"Last name : {last}")
```

```python
parse_and_display(test_name)
```

```python
test_list = ['Grace Flores','John Doe', 'Tom Nook']
```

```python
parse_and_display(test_list[0])
```

```python
parse_and_display(test_list[2])
```

```python
parse_and_display('Brandon Carrillo')
```

```python
def parse_and_display2(name):
    first_space = -1
    
    for i in range(len(name)):
        if name [i] == ' ':
            if first_space == -1:
                first_space = i
            else:
                first = name [:first_space]
                middle = name [first_space+1:i]
                last = name[i+1:]
                print(f"Full name : {name}")
                print(f"Middle name : {middle}")
                print(f"First name : {first}")
                print(f"Last name : {last}")
```

```python
parse_and_display2('Lebron Raymond James')
```
