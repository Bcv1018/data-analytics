## Lab 2. Parsing strings – part codes

```python
code1 = ('ACME:123-L')
code2 = ('DI:12345-M')
code3 = ('ACE:1-12')
```

```python
def get_supplier_code(code):
    for i in range(len(code)):
        if code[i] == ':':
            supplier = code[:i]
            return f'{supplier}'
```

```python
print(get_supplier_code(code1))
print(get_supplier_code(code2))
print(get_supplier_code(code3))
```

```python
def get_product_number(code):
    colon_pos = -1
    dash_pos = -1

    for i in range(len(code)):
        if code[i] == ':':
            colon_pos = i
        elif code[i] == '-':
            dash_pos = i              
    
    product = code[colon_pos+1:dash_pos]
    return f'{product}'
```

```python
print(get_product_number(code1))
print(get_product_number(code2))
print(get_product_number(code3))
```

```python
def get_size(code):
    for i in range(len(code)):
        if code[i] == '-':
            size = code[i+1:]
            return f'{size}'
            
```

```python
print(get_size(code1))
print(get_size(code2))
print(get_size(code3))
```

```python
print(f"For {code1}, the supplier is {get_supplier_code(code1)}. Part number {get_product_number(code1)}# is available in size {get_size(code1)}")
print(f"For {code2}, the supplier is {get_supplier_code(code2)}. Part number {get_product_number(code2)}# is available in size {get_size(code2)}")
print(f"For {code3}, the supplier is {get_supplier_code(code3)}. Part number {get_product_number(code3)}# is available in size {get_size(code3)}")
```
