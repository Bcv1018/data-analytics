## Social Security Tax

```python
def get_soc_sec_tax(gross_pay):
    Sst = gross_pay * 0.062
    return Sst
```

## Medicare Tax

```python
def get_medicare_tax(gross_pay):
    Mt = gross_pay * 0.0145
    return Mt
```

## Federal tax

```python
def get_federal_tax(gross_pay,Wcode):
    if Wcode == 0:
        ft = gross_pay * .023
    elif Wcode == 1:
        ft = gross_pay * .021
    elif Wcode == 2:
        ft = gross_pay * .0195
    elif Wcode == 3:
        ft = gross_pay * .0185
    else:
        ft = gross_pay * (.018 - ((Wcode -4) * .005))
    return f"{ft:.2f}"
        
```

```python
print(get_federal_tax(750,0))
```

```python
print(get_federal_tax(1550,2))
```
