### Working with defining custom functions

```python
def favorite_things(name,movie,music):
    return f'My name is {name}, my favorite movie is {movie}, and my favorite artist right now is {music}'
```

```python
favorite_things('Brandon Carrillo','The Batman','Kodak Black')
```

```python
def why_im_here(purpose):
    return [purpose]
```

```python
why_im_here('I joined the data analytics program at year up united because I want to start the first step in making data analytics my career.')
```

```python
def favorite_place(place,why):
    return f'One of my favorite places is {place} because {why}'
```

```python
favorite_place('Saigon','It is one of my favorite restaurants')
```

```python
favorite_place('my room','it is my safe place and where I spend a lot of my time at')
```
